# **Experiment7**

```python
# Experiment 7 - Cleaning data by Median/Mode Imputation + Standardization + Wrapper(RFE) 13 Features
```

```python
# Create a copy of the cleaned dataset
df_exp = df.copy()

# Define numerical and categorical features
numerical_cols = [
    'Age',
    'Sex',
    'RestBP',
    'Chol',
    'Fbs',
    'RestECG',
    'MaxHR',
    'ExAng',
    'Oldpeak',
    'Slope',
    'Ca'
]

categorical_cols = [
    'ChestPain',
    'Thal'
]

# Impute missing values in numerical features using the median
for col in numerical_cols:
    df_exp[col] = df_exp[col].fillna(df_exp[col].median())

# Impute missing values in categorical features using the mode
for col in categorical_cols:
    df_exp[col] = df_exp[col].fillna(df_exp[col].mode()[0])

# Check the number of missing values after imputation
print("Missing values after imputation:")
print(df_exp.isnull().sum())
```

```python
# One-Hot Encoding
# Convert categorical features into numerical dummy variables
# เปลี่ยน categorical ตัวแปร chestpain,Thal,ADH

df_exp = pd.get_dummies(df_exp, columns=['ChestPain', 'Thal', 'AHD'], drop_first=True) # แปลงข้อมูล categorical เป็น One-hot encoding และตัดคอลัมน์แรกทิ้งเพื่อป้องกัน multicollinearity
df_exp = df_exp.astype(float) # แปลงข้อมูลทุกคอลัมน์ใน DataFrame ให้เป็นชนิดข้อมูลแบบทศนิยม (float)

cols = [c for c in df_exp.columns if c != 'AHD_Yes'] + ['AHD_Yes']  # จัดเรียงลำดับคอลัมน์ใหม่โดยย้ายคอลัมน์ AHD_Yes ไปไว้ท้ายสุด
df_exp = df_exp[cols] # อัปเดต DataFrame ให้ใช้ลำดับคอลัมน์ใหม่ที่จัดเรียงแล้ว

print("Prepared data shape:", df_exp.shape) # แสดงขนาดของ DataFrame (จำนวนแถว, จำนวนคอลัมน์)
# แสดงข้อมูลตัวอย่างของ DataFrame
display(df_exp.head())
```

```python
import numpy as np

# Setting input features X and target output y
X = np.array(df_exp.iloc[:, :-1])
y = np.array(df_exp['AHD_Yes'])

print("X shape:", X.shape)
print("y shape:", y.shape)
n_class = np.unique(y)
```

```python
# Scaling
# Standardization
from sklearn import preprocessing as pre
scaler = pre.StandardScaler().fit(X)
x_scaled = scaler.transform(X)

print("Scaled X shape:", x_scaled.shape)
```

```python
# Split data
import sklearn.model_selection as ms
data = ms.train_test_split(x_scaled, y, 
                           test_size=0.33, 
                           random_state=4, 
                           stratify=y)
x_train = data[0]
x_test = data[1]
y_train = data[2]
y_test = data[3]
```

```python
# Feature Selection using RFE
# Use Logistic Regression to rank and eliminate features recursively

from sklearn.feature_selection import RFE
from sklearn.linear_model import LogisticRegression

# Create Logistic Regression model for RFE
rfe_model = LogisticRegression(
    random_state=0,
    max_iter=1000
)

# Select the Top 13 Features from 16 Features
rfe = RFE(
    estimator=rfe_model,
    n_features_to_select=13,
    step=1
)

# Fit RFE using training data only
rfe.fit(x_train, y_train)

# The target variable (AHD_Yes) is excluded
feature_names = df_exp.iloc[:, :-1].columns

selected_features = feature_names[rfe.support_]

# Display the selected features
print("Selected 13 Features:")
print(list(selected_features))
```

```python
# Get the column positions of the selected features
selected_indices = np.where(rfe.support_)[0]

# Select the same features from training and testing data
x_train_selected = x_train[:, selected_indices]
x_test_selected = x_test[:, selected_indices]

print("X train shape:", x_train_selected.shape)
print("X test shape:", x_test_selected.shape)
```

```python
# Fit Logistic Regression using selected features
clf = LogisticRegression(
    random_state=0,
    max_iter=1000
).fit(x_train_selected, y_train)

# Evaluation --- Experiment 7
print('training performance: ',
      clf.score(x_train_selected, y_train))

print('testing performance: ',
      clf.score(x_test_selected, y_test))
```
